```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def x(s, d_t, T_n): # non-infeted T-cells
    return s - (d_t*T_n)

def f(k, V_i, T_star, T_init,a, t, d): # infected T-cells
    return k*V_i*(T_init+(a*t))-(d*T_star)

def g(n, N, d, T_star, c, V_i): #infectious virus
    return ((1-n)* N * d * T_star) - c*V_i

def y(n, N, d, T_star,c, V_n): #noninfetious virus
    return (n*N*d*T_star) - (c*V_n)

```

```python
def euler(T_n0,T_star0, V_i0, V_n0, T_final, T_init, N, M, k, a, t, d,c,s,d_t,n):
    T_n = np.zeros(M+1)
    T_star = np.zeros(M+1)
    V_i = np.zeros(M+1)
    V_n = np.zeros(M+1)
    T_n[0] = T_n0
    T_star[0] = T_star0
    V_i[0] = V_i0
    V_n[0] = V_n0
    h = T_final/M
    for i in range(M):
        T_n [i+1] = max(T_n[i] + h * x(s, d_t, T_n[i]), 0.0)
        T_star[i+1] = max(T_star[i] + h * f(k, V_i[i], T_star[i], T_init, a, t[i], d), 0.0)
        V_i[i+1] = max(V_i[i] + h * g(n, N, d, T_star[i], c, V_i[i]), 0.0)
        V_n[i+1] = max(V_n[i] + h * y(n, N, d, T_star[i], c, V_n[i]), 0.0)
    return T_n, T_star,V_i, V_n
```

```python
k = 3.43e-5    # clearance parameter, clearance is parameter representing the volume of plasma or blood fully cleared of a drug per unit time
#k = 4.43e-4
delta = 0.43  # death rate of infected cells
#delta = .3
d = delta
N = 480     # virus particles produced by infected T-cells
#c = 2       # The rate at which the immune system clears the infectious virus
c = .5
T_init = 180     # initial uninfected T-cells (T0)
#T_init = 1000
a = 1         # constant rate or T-cell growth rate 
#a = 20
n = 0.2     # a measure of the effectiveness of a protease inhibitor against virus particles
d_t = 0.02
#s = 20    #number of t-cells your body is producing
s= 200
T_final = 15  # x-axis on target graph goes to 15 days
M = 10000
```

```python
#set n = .2 where N = 100
N = 100
T_final = 100
M = 10000
t = np.linspace(0,T_final,M+1)
# T10 = .2, V10= .3, V20 = .4
T_n1 ,T_star1, V_i1, V_n1 =euler(0.3,0.2, 100000*0.8, 100000*0.2,T_final, T_init, N, M, k, a, t, d,c,s, d_t,n=.2)

plt.semilogy(t, T_n1, color="orange", label="healthy T-cells")
plt.semilogy(t, T_star1, color="blue", label="Infected T-cells")
plt.semilogy(t, V_i1, color="red", label="Infectious virus")
plt.semilogy(t, V_n1, color="green", label="Non-infectious virus")

plt.xlabel("Days")
plt.ylabel("Cell and Virus Population")
plt.title("HIV Drug Therapy Model")
plt.legend()


plt.show()
```
